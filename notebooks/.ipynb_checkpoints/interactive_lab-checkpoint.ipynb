# Jury LLM: Multi-Model & Human-in-the-Loop Evaluation System

This notebook serves as the interactive frontend for the Jury System.

```python
import sys
import os
sys.path.append('../')

from dotenv import load_dotenv
load_dotenv('../.env')

import ipywidgets as widgets
from IPython.display import display, Markdown, clear_output
from src.graph import app
from src.utils import parse_json_output
from src.llm_provider import LLMProvider
from src.agents import QUALIFICATION_PROMPT
from dotenv import load_dotenv, find_dotenv
load_dotenv(find_dotenv(), override=True)

llm = LLMProvider()
```

## Step 1: Input Target Text
Please paste the LLM generated text you want to evaluate below. This will be stored globally for the entire session.

```python
# --- Global Variables ---
TARGET_TEXT = "Hi there! 😊
We actually run promotions pretty regularly—especially around holidays, seasonal changes, and special events like Black Friday or our store anniversaries!

Right now, I’d recommend keeping an eye on our homepage or signing up for our newsletter so you don’t miss out on any deals. And if you’re looking for something specific, I can check if there’s an upcoming discount on that item or category!

What kind of products are you interested in? I’d be happy to give you a heads-up! 💡"
HUMAN_COMPETENCY_SCORE = 0
QUALIFICATION_HISTORY = []
MAX_ROUNDS = 3
CURRENT_ROUND = 0

# --- Widgets ---
target_text_input = widgets.Textarea(
    description='Target Text:', 
    placeholder='Paste the LLM generated text here...',
    layout=widgets.Layout(width='80%', height='200px')
)
confirm_text_btn = widgets.Button(description='Confirm & Save Text', button_style='primary')
text_output = widgets.Output()

display(target_text_input, confirm_text_btn, text_output)

def on_confirm_text(b):
    global TARGET_TEXT
    with text_output:
        clear_output()
        val = target_text_input.value.strip()
        if not val:
            print("❌ Error: Text cannot be empty.")
            return
        
        TARGET_TEXT = val
        target_text_input.disabled = True
        confirm_text_btn.disabled = True
        print("✅ Target Text saved successfully! You can now proceed to Step 2.")
        print(f"Preview: {TARGET_TEXT[:100]}...")

confirm_text_btn.on_click(on_confirm_text)
```

## Step 2: Qualification Assessment
The AI will now analyze the text you provided in Step 1 and assess your qualification to evaluate it.

```python
chat_output = widgets.Output(layout={'border': '1px solid #ccc', 'height': '300px', 'overflow_y': 'scroll'})
user_input = widgets.Text(description='Your Answer:', placeholder='Type here...', layout=widgets.Layout(width='80%'))
send_btn = widgets.Button(description='Send', button_style='primary')
stop_btn = widgets.Button(description='Stop/Finish', button_style='warning')
start_qual_btn = widgets.Button(description='Start Assessment', button_style='success')

input_box = widgets.VBox([user_input, widgets.HBox([send_btn, stop_btn])])
qual_container = widgets.VBox([start_qual_btn, chat_output, input_box])

# Initially hide input box until started
input_box.layout.display = 'none'

display(qual_container)

def append_chat(role, text):
    with chat_output:
        if role == "AI":
            display(Markdown(f"**🤖 Examiner:** {text}"))
        else:
            display(Markdown(f"**👤 You:** {text}"))

def run_qualification_step(user_msg=None):
    global CURRENT_ROUND, HUMAN_COMPETENCY_SCORE, QUALIFICATION_HISTORY, TARGET_TEXT
    
    if not TARGET_TEXT:
        with chat_output:
            print("❌ Error: No Target Text found. Please complete Step 1 first.")
        return

    # Add user message to history if exists
    if user_msg:
        QUALIFICATION_HISTORY.append(f"User: {user_msg}")
        
    # Prepare Prompt
    history_str = "\n".join(QUALIFICATION_HISTORY)
    prompt = QUALIFICATION_PROMPT.format(target_text=TARGET_TEXT, history=history_str)
    
    # Call LLM
    # We use a generic model for this, e.g., the judge model or just qwen-flash
    # Note: In a real notebook, this might block UI. For demo, it's fine.
    try:
        response = llm.generate("qwen-flash", [{"role": "user", "content": prompt}])
        data = parse_json_output(response)
    except Exception as e:
        append_chat("AI", f"Error calling LLM: {e}")
        return
    
    status = data.get("status")
    
    if status == "asking" and CURRENT_ROUND < MAX_ROUNDS:
        question = data.get("question")
        QUALIFICATION_HISTORY.append(f"AI: {question}")
        append_chat("AI", question)
        CURRENT_ROUND += 1
    else:
        # Finished or Max Rounds reached
        score = data.get("score", 0)
        reason = data.get("reason", "Evaluation ended.")
        HUMAN_COMPETENCY_SCORE = score
        
        append_chat("AI", f"**Assessment Complete.**\n**Score:** {score}\n**Reason:** {reason}")
        
        # Disable inputs
        user_input.disabled = True
        send_btn.disabled = True
        stop_btn.disabled = True
        
        display(Markdown("✅ **Qualification Step Done. Proceed to Step 3 below.**"))

def on_start_qual(b):
    start_qual_btn.disabled = True
    input_box.layout.display = 'block'
    append_chat("AI", "Analyzing text and your qualification...")
    run_qualification_step()

def on_send(b):
    msg = user_input.value
    if not msg:
        return
    
    append_chat("User", msg)
    user_input.value = ""
    run_qualification_step(user_msg=msg)

def on_stop(b):
    QUALIFICATION_HISTORY.append("User: [STOP SIGNAL]")
    run_qualification_step(user_msg="[STOP SIGNAL]")

start_qual_btn.on_click(on_start_qual)
send_btn.on_click(on_send)
stop_btn.on_click(on_stop)
```

## Step 3: Jury Evaluation
Now that your qualification is established, we proceed to the jury evaluation.

```python
# Widgets for Jury Step
human_score_input = widgets.FloatSlider(value=80, min=0, max=100, description='Your Score:')
human_reason_input = widgets.Textarea(description='Reason:', placeholder='Why did you give this score?')
start_jury_btn = widgets.Button(description='Start Jury Evaluation', button_style='success')
jury_output_area = widgets.Output()

display(human_score_input, human_reason_input, start_jury_btn, jury_output_area)

thread = {"configurable": {"thread_id": "1"}}

def on_start_jury(b):
    with jury_output_area:
        clear_output()
        if not TARGET_TEXT:
            print("Error: No Target Text found. Please complete Step 1.")
            return
            
        print(f"Initializing Jury with Human Competency Score: {HUMAN_COMPETENCY_SCORE}...")
        
        # Map Competency Score (0-100) to Weight (0.5-1.5)
        human_weight = 0.5 + (HUMAN_COMPETENCY_SCORE / 100.0)
        
        initial_state = {
            "topic": TARGET_TEXT,
            "human_bio": "Assessed via Qualification Exam",
            "human_score": human_score_input.value,
            "human_reason": human_reason_input.value,
            "human_weight": human_weight,
            "model_outputs": {},
            "debate_logs": [],
            "votes": {}
        }
        
        for event in app.stream(initial_state, thread, stream_mode="values"):
            if 'debate_logs' in event and event['debate_logs']:
                print("--- Debate Log ---")
                for log in event['debate_logs']:
                    print(log)
            if 'anonymized_reasons' in event:
                print("Ready for Voting...")
                
        # Check state after interruption
        state_snapshot = app.get_state(thread)
        if state_snapshot.next:
            print("System paused for Human Vote.")
            show_voting_ui(state_snapshot.values)

start_jury_btn.on_click(on_start_jury)
```

## Step 4: Voting Interface

```python
vote_widget = widgets.RadioButtons(options=[], description='Best Reason:')
submit_vote_btn = widgets.Button(description='Submit Vote', button_style='info')
vote_output = widgets.Output()

def show_voting_ui(state_values):
    options = state_values['anonymized_reasons']
    display_options = []
    for opt_id, text in options.items():
        display_options.append((f"{opt_id}: {text[:100]}...", opt_id))
    
    vote_widget.options = display_options
    display(vote_widget, submit_vote_btn, vote_output)

def on_vote_submit(b):
    with vote_output:
        selected_option = vote_widget.value
        print(f"You voted for: {selected_option}")
        
        current_votes = app.get_state(thread).values.get('votes', {})
        current_votes['human'] = selected_option
        
        app.update_state(thread, {"votes": current_votes})
        
        print("Resuming execution...")
        for event in app.stream(None, thread, stream_mode="values"):
            if 'final_verdict' in event:
                display(Markdown("### Final Verdict"))
                display(Markdown(event['final_verdict']))

submit_vote_btn.on_click(on_vote_submit)
```
